verifying the diann like format of the  two dataframes

```python
# %%
expected_core_cols = {
    "Precursor.Id",
    "Stripped.Sequence",
    "Charge",
    "Precursor.Normalised",  # ⚠️ Normalised avec s
    "Precursor.PEP",
}

def check_schema(df: pd.DataFrame, label: str):
    print(f"\n=== {label} ===")
    if df is None:
        print("Table is None (file missing or not loaded).")
        return
    cols = set(df.columns)
    print("Columns (sample):", list(sorted(list(cols))[:15]), "...")
    missing = expected_core_cols - cols
    extra = cols - expected_core_cols
    print("Missing core columns:", missing)
    print("Extra columns (beyond core):", list(sorted(list(extra))[:15]), "...")
    print("Number of rows:", len(df))

check_schema(diann_df, "DIA-NN (report.parquet)")
check_schema(mq_df, "MaxQuant (evidence.txt -> DIA-NN-like)")

```

defining a generic analysis function for any dataframe (no matter where the file comes from, once it is adapted, i will paply the same analysis)

```python
# %%
def basic_peptide_qc(df: pd.DataFrame, label: str, pep_threshold: float = 0.01):
    """
    Applique une QA simple sur un tableau DIA-NN-like :
    - filtre sur Precursor.PEP
    - vérifie que Stripped.Sequence et Charge sont bien renseignés
    - regarde la distribution des intensités Precursor.Normalised
    """
    if df is None:
        print(f"\n[{label}] No data (df is None). Skipping.")
        return

    print(f"\n===== Basic QC for {label} =====")
    print("Initial rows:", len(df))

    # filtrage PEP
    if "Precursor.PEP" in df.columns:
        df_filt = df[df["Precursor.PEP"] <= pep_threshold].copy()
        print(f"Rows after PEP <= {pep_threshold}: {len(df_filt)}")
    else:
        print("WARNING: 'Precursor.PEP' not found, skipping PEP filtering.")
        df_filt = df.copy()

    # séquences non vides et charge positive
    if "Stripped.Sequence" in df_filt.columns:
        non_empty_seq = (df_filt["Stripped.Sequence"].astype(str).str.len() > 0).sum()
        print("Non-empty sequences:", non_empty_seq)
    if "Charge" in df_filt.columns:
        positive_charge = (df_filt["Charge"] > 0).sum()
        print("Charge > 0:", positive_charge)

    # distribution des intensités
    if "Precursor.Normalised" in df_filt.columns:
        valid_int = df_filt["Precursor.Normalised"] > 0
        n_valid = valid_int.sum()
        print("Rows with Precursor.Normalised > 0:", n_valid)

        if n_valid > 0:
            log_int = np.log10(df_filt.loc[valid_int, "Precursor.Normalised"])
            print("log10(Precursor.Normalised):",
                  "median =", float(log_int.median()),
                  ", mean =", float(log_int.mean()))

            plt.figure(figsize=(6, 4))
            plt.hist(log_int, bins=50)
            plt.xlabel("log10 Precursor.Normalised")
            plt.ylabel("Count")
            plt.title(f"Intensity distribution for {label}")
            plt.grid(True)
            plt.show()
        else:
            print("No positive Precursor.Normalised values to plot.")
    else:
        print("WARNING: 'Precursor.Normalised' missing, no intensity plot.")
```

applying the same to diann and maxquant

```python
# %%
basic_peptide_qc(diann_df, "DIA-NN report.parquet")
basic_peptide_qc(mq_df,   "MaxQuant evidence.txt (converted)")
```

moving on to actual downstream analysis

helper: finding the protein column

```python
# %%
def get_protein_column(df: pd.DataFrame) -> str | None:
    """
    Essaie de deviner quelle colonne contient l'identifiant de protéine.
    On teste plusieurs noms standards.
    """
    if df is None:
        return None

    candidates = [
        "Protein.Names",
        "Protein.Group",
        "Protein.Ids",
        "Leading.Proteins",
        "Protein",
    ]
    for col in candidates:
        if col in df.columns:
            return col
    return None
```

intensity per protein

```python
# %%
def summarize_intensity_by_protein(
    df: pd.DataFrame,
    label: str,
    agg: str = "sum",
    top_n: int = 20,
):
    """
    Agrège les intensités Precursor.Normalised au niveau protéine et
    affiche les top_n plus intenses.
    """
    if df is None:
        print(f"\n[{label}] No data (df is None). Skipping.")
        return

    protein_col = get_protein_column(df)
    if protein_col is None:
        print(f"\n[{label}] No protein column found. Skipping protein-level summary.")
        return

    if "Precursor.Normalised" not in df.columns:
        print(f"\n[{label}] 'Precursor.Normalised' not found. Skipping.")
        return

    print(f"\n===== Protein-level intensity for {label} =====")
    print("Using protein column:", protein_col)

    # On garde seulement les lignes avec intensité positive
    df_valid = df[df["Precursor.Normalised"] > 0].copy()
    if df_valid.empty:
        print("No positive Precursor.Normalised values. Nothing to summarize.")
        return

    if agg == "sum":
        prot_int = (
            df_valid.groupby(protein_col)["Precursor.Normalised"]
            .sum()
            .sort_values(ascending=False)
        )
    elif agg == "mean":
        prot_int = (
            df_valid.groupby(protein_col)["Precursor.Normalised"]
            .mean()
            .sort_values(ascending=False)
        )
    else:
        raise ValueError("agg must be 'sum' or 'mean'")

    print(f"Number of proteins with non-zero intensity: {len(prot_int)}")

    top = prot_int.head(top_n)
    display(top.to_frame(name=f"Precursor.Normalised_{agg}").head(top_n))

    # Petit barplot pour les top_n
    plt.figure(figsize=(10, 4))
    top.plot(kind="bar")
    plt.ylabel(f"{agg} Precursor.Normalised")
    plt.title(f"Top {top_n} proteins by {agg} intensity ({label})")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

histogramm of peptide length

```python
# %%
def plot_peptide_length_distribution(
    df: pd.DataFrame,
    label: str,
    max_length: int = 60,
):
    """
    Trace un histogramme des longueurs de peptides (colonne 'length').
    """
    if df is None:
        print(f"\n[{label}] No data (df is None). Skipping.")
        return

    if "length" not in df.columns:
        print(f"\n[{label}] 'length' column not found. Skipping length histogram.")
        return

    lengths = df["length"].dropna()
    lengths = lengths[lengths > 0]

    if lengths.empty:
        print(f"\n[{label}] No positive peptide lengths. Skipping.")
        return

    # On tronque pour éviter quelques outliers bizarres éventuels
    lengths = lengths[lengths <= max_length]

    print(f"\n===== Peptide length distribution for {label} =====")
    print("Number of peptides:", len(lengths))
    print("Median length:", float(lengths.median()))
    print("Mean length:", float(lengths.mean()))

    plt.figure(figsize=(6, 4))
    plt.hist(lengths, bins=range(1, max_length + 2))
    plt.xlabel("Peptide length (aa)")
    plt.ylabel("Count")
    plt.title(f"Peptide length distribution ({label})")
    plt.grid(True)
    plt.show()
```

distribution of missed  cleavages

```python
# %%
def plot_missed_cleavages_distribution(df: pd.DataFrame, label: str):
    """
    Regarde la distribution de 'missed_cleavages' si la colonne existe.
    """
    if df is None:
        print(f"\n[{label}] No data (df is None). Skipping.")
        return

    col = "missed_cleavages"
    if col not in df.columns:
        print(f"\n[{label}] '{col}' column not found. Skipping.")
        return

    mc = df[col].dropna().astype(int)

    if mc.empty:
        print(f"\n[{label}] No missed_cleavages data. Skipping.")
        return

    print(f"\n===== Missed cleavages distribution for {label} =====")
    print("Value counts:")
    display(mc.value_counts().sort_index())

    plt.figure(figsize=(6, 4))
    mc.value_counts().sort_index().plot(kind="bar")
    plt.xlabel("Missed cleavages")
    plt.ylabel("Count")
    plt.title(f"Missed cleavages distribution ({label})")
    plt.grid(True)
    plt.show()
```

applying these analyses to diann and maxquant

```python
# %%
# DIA-NN (report.parquet)
summarize_intensity_by_protein(diann_df, "DIA-NN report.parquet", agg="sum", top_n=20)
plot_peptide_length_distribution(diann_df, "DIA-NN report.parquet", max_length=60)
plot_missed_cleavages_distribution(diann_df, "DIA-NN report.parquet")

# %%
# MaxQuant (evidence.txt -> format DIA-NN-like)
summarize_intensity_by_protein(mq_df, "MaxQuant evidence.txt (converted)", agg="sum", top_n=20)
plot_peptide_length_distribution(mq_df, "MaxQuant evidence.txt (converted)", max_length=60)
plot_missed_cleavages_distribution(mq_df, "MaxQuant evidence.txt (converted)")
```
